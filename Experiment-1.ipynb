```
from transformers import pipeline

# Load GPT-2 text generation pipeline
generator = pipeline(
    "text-generation",
    model="gpt2"
)

headline = input("Enter News Headline: ")

prompt = f"Write a professional news article about: {headline}\n\nArticle:\n"

result = generator(
    prompt,
    max_length=180,
    num_return_sequences=1,
    temperature=0.8,
    do_sample=True
)

print("\nGenerated Article:\n")
print(result[0]["generated_text"])
```

```
from transformers import pipeline

# Sentiment analysis
sentiment = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# Text generation
generator = pipeline(
    "text-generation",
    model="google/flan-t5-large"
)
user_input = input("Enter headline or topic: ")

analysis = sentiment(user_input)

print(analysis)
label = analysis[0]["label"]

if label == "POSITIVE":
    prompt = f"""
    You are a journalist.

    Write an optimistic news article based on:

    {user_input}

    Highlight achievements and positive outcomes.
    """

elif label == "NEGATIVE":
    prompt = f"""
    You are an investigative journalist.

    Write a balanced news article based on:

    {user_input}

    Discuss challenges, causes, and possible solutions.
    """

elif label == "NEUTRAL":
    prompt = f"""
    Write a neutral professional news article about:

    {user_input}"""
else:
    prompt = f"""
    Write a professional news article about:

    {user_input}"""

result = generator(
    prompt,
    max_new_tokens=500,
    temperature=0.8
)

print(result[0]["generated_text"])
```
